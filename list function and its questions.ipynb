```python
# list of same datatypes 
numbers = [10,20,30,40,50]
print(numbers)
print(numbers[2])
print(len(numbers))
```

```python
# performing the manipulation 
numbers.insert(1,200)
print(numbers)
numbers.append(600)
print(numbers)
numbers.insert(2,300)
print(numbers)
numbers.remove(40)
print(numbers)
numbers.pop()
print(numbers)
```

```python
# ascending orrdeer
numbers.sort()
print(numbers)
# descending order
numbers.sort(reverse=True)
print(numbers)
```
