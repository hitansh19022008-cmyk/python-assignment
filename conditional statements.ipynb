```python
# check whether person is eligible to vote or notr using the if statement 
age = 20 
if (age>=18):
    print("Eligible to vote")
if (age<=18):
    print("Not eligible to vote")
```

```python
# using the elif statement 
# elif will check statement only if if statement is false 
Light = "Green"
if(Light == "Red"):
    print("stop")
elif(Light == "Green"):
    print("Go")
```

```python
# if statement is true then elif statement is not of use 
Light = "Red"
if(Light == "Red"):
    print("stop")
elif(Light == "Yellow"):
    print("Ready")
```

```python
# both if statements can un togther 
age = 18
if (age>=18):
    print("Eligible to vote")
if (age<19):
    print("Under age")
```

```python
# Using the else statement 
# it is used in the last
# not check the statement 
# to find the largest no 
Number1 = int(input("Enter first number:"))
Number2 = int(input("Enter second number:"))
Number3 = int(input("Enter third number:"))
# to find largest number 
if (Number1>Number2 and Number1>Number3):
    print("Number1 is largest number")
elif (Number2>Number1 and Number2>Number3):
    print("Number2 is largest number")
elif (Number3>Number1 and Number3>Number2):
    print("Number3 is largest number")
else:
    print("All numbers are distinct")
```

```python
# if nesting statements
age = int(input("Enter your age"))
if(age>=18):
    if(age>=75):
        print("Cannot drive")
    else:
        print("Can drive")
```

```python
# to check the number is even or odd 
number = int(input("Enter your number:"))
if(number%2 == 0):
    print("number is even")
else:
    print("number is odd")
```

```python
# check whether no is multiple of 7 or not 
number = int(input("Enter your number :"))
if(number%7 == 0):
    print("Multiple of 7")
else:
    print("Not multiple of 7")
```

```python
# to chech the grade of student 
marks = int(input("Enter your marks obtained:"))
if(90<=marks<=100):
    print("Grade == A+")
elif(80<=marks<=89):
    print("Grade == A")
elif(70<=marks<=79):
    print("Grade == B")
elif(60<=marks<=69):
    print("Grade == C")
elif(50<=marks<=59):
    print("Grade == D")
elif(40<=marks<=49):
    print("Grade == E")
elif(30<=marks<=39):
    print("Grade == F")
else:
    print("Grade is not assigned to student ")
    
```
