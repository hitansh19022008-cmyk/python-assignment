```python
# dictionary is unordered , mutuable, do not allow duplicates
# formed using the parentheses{}
info = {"Name":"Hitansh","Surname":"Jain","Age":18,"PRN":1272260661}
print(info)
```

```python
# Accesing the element 
print(info["Name"])
```

```python
print(info["Age"])
```

```python
# adding the element
info["Class"] = "CSE (CLOUD COMPUTING)"
print(info)
```

```python
info["Division"] = 14
print(info)
```

```python
# updating the element 
info.update({"PRN":1272260661234,"Class":"B.TECH.CSE(CLOUD COMPUTING)"})
print(info)
```

```python
# to see all keys in dictionary 
print(info.keys())
```

```python
# to see all values in dictionary 
print(info.values())
```

```python
# to see all items in dictionary 
print(info.items())
```

```python
# delete in dictionary
print(info.pop("Division"))
```
