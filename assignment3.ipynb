```python
# program to check right angle triangle
def check_triangle(k,l,m):
    if k*k==l*l+m*m:
        print(" It is a right angle triangle")
    elif l*l==k*k+m*m:
        print("It is a right angle triangle")
    elif m*m==k*k+l*l:
        print("It is a right angle triangle")
    else:
        print("It is not a  right angled triangle")
k=int(input("Enter the first side:"))
l=int(input("Enter the second side:"))
m=int(input("Enter the third side:"))
check_triangle(k,l,m)
    
```
