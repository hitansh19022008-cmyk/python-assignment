```python
# To count the number of students with the Grade "A" in the following tuple.
# also store above values in list & sort them from "A' to "B"
Grade = ("C","D","A","A","B","B","A")
print(Grade)
print(Grade.count("A"))
# to make list
b = list(Grade)
print(b)
b.sort()
print(b)
```

```python
# tuple is same like list but immutable
# ordered , store differnt data types 
# stored using the rou d brackett
rank = (12722,147224,15678)
print(rank)
```

```python
print(rank[1])
```

```python
# to check the length
print(len(rank))
```

```python
# to check min and max in tuple
print(max(rank))
print(min(rank))
```

```python
# to count the elemnt
print(rank.count(12722))
```

```python
# to add element in tuple 
rank = (12722,147224,13456) 
print(rank)
a = list(rank)
print(a)
a.extend([15678,1982])
b = tuple(a)
print(b)
```
